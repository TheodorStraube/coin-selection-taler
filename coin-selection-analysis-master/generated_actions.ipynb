```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy.optimize import curve_fit

def load_generated_actions():
    data_dir = '../coin-selection-c-master/simulation/users'
    files = [os.path.join(data_dir, f) for f in os.listdir(data_dir) if f.endswith('.csv')]
    
    user_names = ["STUDENT", "STUDENT_STATIC", "BUSINESS_OWNER", "RETIRED", "FAMILY", "FREELANCER", "TEACHER", "ARTIST", "BALANCED_ARTIST"]
    op_names = ["DEPOSIT_OP", "WITHDRAW_OP", "REFUND_OP", "REFRESH_OP", "WIRE_OP", "CLOSE_OP"]
    strategy_names = ["MAX_BILLS",
            "MIN_BILLS",
            "CLOSEST_TO_EXPIRE_MIN_BILLS",
            "CLOSEST_TO_EXPIRE_MAX_BILLS",
            "MAX_BILLS_TIME_TO_EXPIRE_WEIGHTED",
            "RANDOM",
            "EVEN_FROM_MIN_TO_MAX",
            "EVEN_FROM_MAX_TO_MIN",
            "GREEDY_MIN_TO_MAX",
            "GREEDY_MIN_TO_MAX_FIX",
            "GREEDY_MAX_TO_MIN_FIX",
            "CALL_EXTERNAL",
            "WALLET_CORE"]
    
    if not files:
        raise IOError("No files found")
    
    def get_first_line(path):
        with open(path, 'r') as inp:
            return inp.readline().strip().split(',')
    
    columns = ["OperationId", "Amount", "Time"]
    header_columns = ["UserName", "UserTypeId", "StrategyId", "UserId"]
    dfs = []
    
    for path in files:
        csv = pd.read_csv(path, dtype=pd.Int64Dtype(), names=columns, skiprows=[0])
        
        user_name, user_type, strategy, user_id = get_first_line(path)
        csv["UserType"] = user_names[int(user_type)]
        csv["Strategy"] = strategy_names[int(strategy)]
        csv["Operation"] = csv["OperationId"].map({i: op_name for i, op_name in enumerate(op_names)})
        
        dfs.append(csv)
    
    return pd.concat(dfs)
```
